```python
# fib
import time

def fib(n):
    if n == 1 or n == 0:
        return 1
    else:
        return fib(n-1) + fib(n-2)
```

```python
start = time.time()
fib(5)
print(time.time() - start)
```

Code works but highly inefficient and exponential approach(in notebook) ; as the input increases the time to execute also increases to tackle this use MEMOIZATION.

```python
# Using Memoization
# fib
import time

def fib(n, d):
    if n in d:
        return d[n]
    else:
        d[n] = fib(n-1, d) + fib(n-2, d)
        return d[n]
```

```python
start = time.time()
d = {0: 1, 1: 1}
fib(39, d)
print(time.time() - start)
```

```python
start = time.time()
d = {0: 1, 1: 1}
fib(45, d)
print(time.time() - start)
```

Notice how the time reduced for execution
